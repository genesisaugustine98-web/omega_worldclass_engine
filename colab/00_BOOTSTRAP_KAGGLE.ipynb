# OMEGA Kaggle fallback bootstrap
Academic research only. Configure the repository URL and permitted dataset access before running.

```python
import os, subprocess, sys
PROJECT = '/kaggle/working/omega_worldclass_engine'
REPO = os.environ.get('OMEGA_REPO_URL', '')
if not os.path.exists(PROJECT):
    if not REPO:
        raise RuntimeError('Set OMEGA_REPO_URL in notebook metadata or environment')
    subprocess.run(['git', 'clone', '--depth', '1', REPO, PROJECT], check=True)
os.chdir(PROJECT)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
os.environ.setdefault('OMEGA_DATA_ROOT', '/kaggle/working/omega_data')
os.environ.setdefault('OMEGA_RUN_ROOT', '/kaggle/working/omega_runs')
```

```python
import json, sys
sys.path.insert(0, PROJECT)
from omega.bootstrap import bootstrap
print(json.dumps(bootstrap(PROJECT, required_free_bytes=500_000_000).as_dict(), indent=2))
```
