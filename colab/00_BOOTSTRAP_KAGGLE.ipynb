# OMEGA Kaggle bootstrap
Academic research only. This notebook clones the repository, installs dependencies, and reports which data-provider credentials are available in Kaggle Secrets. It performs no download until a provider is explicitly configured.

Kaggle free tier provides ~30 GPU hours/week. Note that `/kaggle/working` is **ephemeral** — it resets each session — so use a Kaggle Dataset (or re-run the bootstrap) to persist `OMEGA_DATA_ROOT` across sessions.

```python
import os, subprocess, sys
PROJECT = '/kaggle/working/omega_worldclass_engine'
# Your repository URL; edit if you host it elsewhere.
REPO = os.environ.get('OMEGA_REPO_URL', 'https://github.com/genesisaugustine98-web/omega_worldclass_engine.git')
if not os.path.exists(PROJECT):
    subprocess.run(['git', 'clone', '--depth', '1', REPO, PROJECT], check=True)
os.chdir(PROJECT)
# Core deps plus optional ML models (lightgbm/catboost). Skip the extras if
# you only need the sklearn models.
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt', 'lightgbm', 'catboost'], check=True)
os.environ.setdefault('OMEGA_DATA_ROOT', '/kaggle/working/omega_data')
os.environ.setdefault('OMEGA_RUN_ROOT', '/kaggle/working/omega_runs')
```

## Provider credentials

Add secrets in Kaggle via **Add-ons → Secrets** (the key icon in the right panel) with a linked scope for this notebook. Add any of these names — never commit them:

- `OMEGA_POLYGON_API_KEY` — free Polygon key (daily-resampled M30, opt-in)
