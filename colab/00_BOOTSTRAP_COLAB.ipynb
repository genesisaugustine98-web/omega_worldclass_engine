```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import os, subprocess, sys
PROJECT = '/content/omega_worldclass_engine'
REPO = os.environ.get('OMEGA_REPO_URL', '')
if not os.path.exists(PROJECT):
    if not REPO:
        raise RuntimeError('Set OMEGA_REPO_URL to a repository URL before running bootstrap')
    subprocess.run(['git', 'clone', '--depth', '1', REPO, PROJECT], check=True)
os.chdir(PROJECT)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
os.environ.setdefault('OMEGA_DATA_ROOT', '/content/drive/MyDrive/OMEGA_DATA_v2')
os.environ.setdefault('OMEGA_RUN_ROOT', '/content/drive/MyDrive/OMEGA_RUNS_v2')
```

```python
import json, sys
sys.path.insert(0, PROJECT)
from omega.bootstrap import bootstrap
print(json.dumps(bootstrap(PROJECT).as_dict(), indent=2))
```
