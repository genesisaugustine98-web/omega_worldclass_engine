# OMEGA Colab bootstrap
Academic research only. This notebook clones the repository, installs dependencies, mounts Drive, and reports which data-provider credentials are available. It performs no download until a provider is explicitly configured.


```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import os, subprocess, sys
PROJECT = '/content/omega_worldclass_engine'
# Your repository URL; edit if you host it elsewhere.
REPO = os.environ.get('OMEGA_REPO_URL', 'https://github.com/genesisaugustine98-web/omega_worldclass_engine.git')
if not os.path.exists(PROJECT):
    subprocess.run(['git', 'clone', '--depth', '1', REPO, PROJECT], check=True)
os.chdir(PROJECT)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
os.environ.setdefault('OMEGA_DATA_ROOT', '/content/drive/MyDrive/OMEGA_DATA_v2')
os.environ.setdefault('OMEGA_RUN_ROOT', '/content/drive/MyDrive/OMEGA_RUNS_v2')
```

## Provider credentials

`load_platform_secrets` reads secrets from Colab's `userdata` store (🔑 key icon in the left panel). Add any of the names below there — never commit them:

- `OMEGA_TWELVEDATA_API_KEY` — free Twelve Data key (recommended, native M30)
- `OMEGA_POLYGON_API_KEY` — free Polygon key (daily-resampled M30, opt-in)
- `OMEGA_OANDA_TOKEN` — OANDA (not available in every jurisdiction)
- `OMEGA_TELEGRAM_BOT_TOKEN` + `OMEGA_TELEGRAM_CHAT_ID` — optional notifications
